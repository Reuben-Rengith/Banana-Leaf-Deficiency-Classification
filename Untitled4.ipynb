```python
VALID_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def is_image_file(fname: str) -> bool:
    return Path(fname).suffix.lower() in VALID_EXTS

def count_images_per_class(dataset_dir: str, classes: list[str]) -> dict[str, int]:
    counts = {}
    for c in classes:
        cdir = os.path.join(dataset_dir, c)
        if not os.path.isdir(cdir):
            counts[c] = 0
            continue
        counts[c] = sum(is_image_file(f) for f in os.listdir(cdir))
    return counts

def validate_dataset(dataset_dir: str, classes: list[str]) -> None:
    if not os.path.exists(dataset_dir):
        raise FileNotFoundError(f"[ERROR] Dataset root not found: {dataset_dir}")

    missing = [c for c in classes if not os.path.isdir(os.path.join(dataset_dir, c))]
    if missing:
        raise FileNotFoundError(f"[ERROR] Missing class folders: {missing}")

    counts = count_images_per_class(dataset_dir, classes)

    print("\n[DEBUG] Image counts per class:")
    for c in classes:
        print(f"  - {c}: {counts[c]}")

    empty = [c for c in classes if counts[c] == 0]
    if empty:
        raise ValueError(f"[ERROR] Empty class folders found: {empty}")

    print("\n[OK] Dataset sanity check passed.")

def build_generators(dataset_dir: str, classes: list[str], img_size=(224,224), batch_size=32, val_split=0.2, seed=42):
    train_datagen = ImageDataGenerator(
        rescale=1/255.0,
        validation_split=val_split,
        rotation_range=25,
        zoom_range=0.25,
        brightness_range=(0.7, 1.3),
        width_shift_range=0.15,
        height_shift_range=0.15,
        horizontal_flip=True,
        fill_mode="nearest"
    )

    val_datagen = ImageDataGenerator(
        rescale=1/255.0,
        validation_split=val_split
    )

    train_gen = train_datagen.flow_from_directory(
        dataset_dir,
        target_size=img_size,
        batch_size=batch_size,
        class_mode="categorical",
        classes=classes,       # enforces consistent indexing
        subset="training",
        seed=seed,
        shuffle=True,
        follow_links=True
    )

    val_gen = val_datagen.flow_from_directory(
        dataset_dir,
        target_size=img_size,
        batch_size=batch_size,
        class_mode="categorical",
        classes=classes,       # enforces consistent indexing
        subset="validation",
        seed=seed,
        shuffle=False,        # stable evaluation order
        follow_links=True
    )

    print("\n[DEBUG] Generator samples:")
    print("  - train samples:", train_gen.samples)
    print("  - val samples:", val_gen.samples)
    print("  - class_indices:", train_gen.class_indices)

    if train_gen.samples == 0:
        raise ValueError("[ERROR] 0 training images found. Check dataset path/extensions/classes.")
    if val_gen.samples == 0:
        raise ValueError("[ERROR] 0 validation images found. Lower val_split or verify dataset size.")

    # Quick smoke test shapes
    x_batch, y_batch = next(train_gen)
    print("\n[DEBUG] One training batch shape:")
    print("  - x_batch:", x_batch.shape)   # (batch, H, W, C)
    print("  - y_batch:", y_batch.shape)   # (batch, num_classes)

    return train_gen, val_gen
```

```python
def categorical_focal_loss(gamma=2.0, alpha=0.25):
    # Works with one-hot y_true and softmax y_pred
    def loss(y_true, y_pred):
        eps = tf.keras.backend.epsilon()
        y_pred = tf.clip_by_value(y_pred, eps, 1.0 - eps)
        ce = -y_true * tf.math.log(y_pred)                     # (batch, C)
        weight = alpha * tf.pow(1.0 - y_pred, gamma)           # (batch, C)
        fl = weight * ce
        return tf.reduce_mean(tf.reduce_sum(fl, axis=1))
    return loss

def unfreeze_last_n(base_model, n=30):
    # Keep BatchNorm layers stable during fine-tuning
    for l in base_model.layers[:-n]:
        l.trainable = False
    for l in base_model.layers[-n:]:
        if not isinstance(l, tf.keras.layers.BatchNormalization):
            l.trainable = True

def get_callbacks(prefix):
    return [
        tf.keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=6, restore_best_weights=True, verbose=1
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss", factor=0.5, patience=3, min_lr=1e-7, verbose=1
        ),
        tf.keras.callbacks.ModelCheckpoint(
            filepath=os.path.join(RESULTS_DIR, f"{prefix}_best.h5"),
            monitor="val_loss", save_best_only=True, verbose=1
        )
    ]

def build_mobilenetv2(num_classes=3, input_shape=(224,224,3)):
    # Transfer learning speeds convergence and improves generalization.
    base = tf.keras.applications.MobileNetV2(
        include_top=False, weights="imagenet", input_shape=input_shape
    )
    base.trainable = False

    inp = layers.Input(shape=input_shape)
    x = base(inp, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.35)(x)
    out = layers.Dense(num_classes, activation="softmax")(x)

    model = Model(inp, out, name="MobileNetV2_3class")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model, base

def build_efficientnetb0(num_classes=5, input_shape=(224,224,3), lr=1e-5):
    base = tf.keras.applications.EfficientNetB0(
        include_top=False, weights="imagenet", input_shape=input_shape
    )
    base.trainable = False

    inp = layers.Input(shape=input_shape)
    x = base(inp, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.45)(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.30)(x)
    out = layers.Dense(num_classes, activation="softmax")(x)

    model = Model(inp, out, name="EfficientNetB0_5class")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(lr),
        loss=categorical_focal_loss(gamma=2.0, alpha=0.25),
        metrics=["accuracy"]
    )
    return model, base

def compute_class_weights(train_gen):
    # train_gen.classes contains integer class ids aligned to train_gen.class_indices
    y = train_gen.classes
    cls = np.unique(y)
    w = compute_class_weight(class_weight="balanced", classes=cls, y=y)
    weights = {int(c): float(v) for c, v in zip(cls, w)}
    print("[DEBUG] class_weights:", weights)
    return weights
```

```python
model1, base1 = build_mobilenetv2(num_classes=len(CLASSES_3), input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3))
model1.summary()

hist1 = model1.fit(
    train3,
    validation_data=val3,
    epochs=20,
    callbacks=get_callbacks("mobilenet3"),
    verbose=1
)

model1.save(os.path.join(RESULTS_DIR, "mobilenet3_final.h5"))
print("\n[OK] Saved:", os.path.join(RESULTS_DIR, "mobilenet3_final.h5"))
```

```python
from sklearn.metrics import classification_report, confusion_matrix, f1_score, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os

def plot_history(history, prefix):
    h = history.history
    # Accuracy
    if "accuracy" in h:
        plt.figure(figsize=(7,5))
        plt.plot(h.get("accuracy", []), label="train_acc")
        plt.plot(h.get("val_accuracy", []), label="val_acc")
        plt.title(f"{prefix} Accuracy")
        plt.xlabel("Epoch"); plt.ylabel("Accuracy")
        plt.legend(); plt.tight_layout()
        plt.savefig(os.path.join(RESULTS_DIR, f"{prefix}_accuracy.png"), dpi=200)
        plt.show()

    # Loss
    if "loss" in h:
        plt.figure(figsize=(7,5))
        plt.plot(h.get("loss", []), label="train_loss")
        plt.plot(h.get("val_loss", []), label="val_loss")
        plt.title(f"{prefix} Loss")
        plt.xlabel("Epoch"); plt.ylabel("Loss")
        plt.legend(); plt.tight_layout()
        plt.savefig(os.path.join(RESULTS_DIR, f"{prefix}_loss.png"), dpi=200)
        plt.show()

def evaluate_model(model, val_gen, class_names, prefix):
    y_true = val_gen.classes
    y_prob = model.predict(val_gen, verbose=0)
    y_pred = np.argmax(y_prob, axis=1)

    acc = accuracy_score(y_true, y_pred)
    macro_f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)
    report_dict = classification_report(y_true, y_pred, target_names=class_names, output_dict=True, zero_division=0)
    cm = confusion_matrix(y_true, y_pred)

    # Per-class recall
    per_class_recall = {c: report_dict[c]["recall"] for c in class_names}
    # Per-class accuracy (from confusion matrix)
    per_class_acc = cm.diagonal() / np.clip(cm.sum(axis=1), 1, None)
    per_class_accuracy = {c: float(per_class_acc[i]) for i, c in enumerate(class_names)}

    # Save text report
    report_path = os.path.join(RESULTS_DIR, f"{prefix}_classification_report.txt")
    with open(report_path, "w") as f:
        f.write(classification_report(y_true, y_pred, target_names=class_names, zero_division=0))
        f.write("\n\nMacro-F1: {:.4f}".format(macro_f1))
        f.write("\nAccuracy: {:.4f}".format(acc))
        f.write("\n\nPer-class recall:\n")
        for c in class_names:
            f.write(f"{c}: {per_class_recall[c]:.4f}\n")

    # Confusion matrix heatmap
    plt.figure(figsize=(8,6))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names)
    plt.title(f"{prefix} Confusion Matrix")
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, f"{prefix}_confusion_matrix.png"), dpi=200)
    plt.show()

    return {
        "accuracy": float(acc),
        "macro_f1": float(macro_f1),
        "per_class_recall": per_class_recall,
        "per_class_accuracy": per_class_accuracy,
        "y_true": y_true,
        "y_pred": y_pred,
        "y_prob": y_prob,
        "cm": cm
    }

# Plot learning curves for MobileNet
plot_history(hist1, "mobilenet3")

# Evaluate MobileNet on its validation generator
m1_metrics = evaluate_model(model1, val3, CLASSES_3, "mobilenet3")
print("\n[Model 1] Accuracy:", m1_metrics["accuracy"])
print("[Model 1] Macro-F1:", m1_metrics["macro_f1"])
```

```python
# Rebuild generators with milder augmentation first (same classes_5)
train5, val5 = build_generators(
    DATASET_ROOT, CLASSES_5,
    img_size=(260, 260), batch_size=24,   # smaller batch for larger image
    val_split=VAL_SPLIT, seed=SEED
)

cw5 = compute_class_weights(train5)

# Rebuild model with updated focal
def categorical_focal_loss(gamma=1.5, alpha=0.5):
    def loss(y_true, y_pred):
        eps = tf.keras.backend.epsilon()
        y_pred = tf.clip_by_value(y_pred, eps, 1.0 - eps)
        ce = -y_true * tf.math.log(y_pred)
        w = alpha * tf.pow(1.0 - y_pred, gamma)
        return tf.reduce_mean(tf.reduce_sum(w * ce, axis=1))
    return loss

model2, base2 = build_efficientnetb0(num_classes=len(CLASSES_5), input_shape=(260,260,3), lr=3e-4)
model2.compile(optimizer=tf.keras.optimizers.Adam(3e-4),
               loss=categorical_focal_loss(1.5, 0.5),
               metrics=["accuracy"])

# Phase A
hA = model2.fit(train5, validation_data=val5, epochs=8, class_weight=cw5,
                callbacks=get_callbacks("eff5_A"), verbose=1)

# Phase B
unfreeze_last_n(base2, n=40)
model2.compile(optimizer=tf.keras.optimizers.Adam(1e-5),
               loss=categorical_focal_loss(1.5, 0.5),
               metrics=["accuracy"])
hB = model2.fit(train5, validation_data=val5, initial_epoch=len(hA.history["loss"]),
                epochs=20, class_weight=cw5, callbacks=get_callbacks("eff5_B"), verbose=1)

# Phase C
unfreeze_last_n(base2, n=80)
model2.compile(optimizer=tf.keras.optimizers.Adam(5e-6),
               loss=categorical_focal_loss(1.5, 0.5),
               metrics=["accuracy"])
hC = model2.fit(train5, validation_data=val5, initial_epoch=20,
                epochs=30, class_weight=cw5, callbacks=get_callbacks("eff5_C"), verbose=1)

model2.save("/content/results/efficientnet5_final_v2.h5")
print("Saved /content/results/efficientnet5_final_v2.h5")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, f1_score, precision_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Ensure these exist from previous cells:
# model1 (MobileNet 3-class), model2 (EfficientNet 5-class), val5, CLASSES_3, CLASSES_5

def predict_paths(model, paths, img_size):
    out = []
    for p in paths:
        im = keras_image.load_img(p, target_size=img_size)
        arr = keras_image.img_to_array(im) / 255.0
        arr = np.expand_dims(arr, axis=0)
        out.append(model.predict(arr, verbose=0)[0])
    return np.array(out)

val_paths = val5.filepaths
y_true = val5.classes

# IMPORTANT: sizes must match how each model was trained
MOB_SIZE = (224, 224)     # change if you trained MobileNet with different size
EFF_SIZE = (260, 260)     # change to (224,224) if your EfficientNet used 224

mob_probs = predict_paths(model1, val_paths, MOB_SIZE)   # shape: [N,3]
eff_probs = predict_paths(model2, val_paths, EFF_SIZE)   # shape: [N,5]

print("mob_probs:", mob_probs.shape, "eff_probs:", eff_probs.shape, "y_true:", y_true.shape)
print("CLASSES_3:", CLASSES_3)
print("CLASSES_5:", CLASSES_5)
```

```python
shared = [c for c in CLASSES_3 if c in CLASSES_5]
shared_eff_idx = [CLASSES_5.index(c) for c in shared]
shared_mob_idx = [CLASSES_3.index(c) for c in shared]

def ensemble_probs_5class(mob_probs, eff_probs, alpha):
    """
    alpha = weight for MobileNet on shared classes.
    final shared prob = alpha*mob + (1-alpha)*eff
    iron/zinc stay from EfficientNet.
    """
    final = eff_probs.copy()

    for mi, ei in zip(shared_mob_idx, shared_eff_idx):
        final[:, ei] = alpha * mob_probs[:, mi] + (1.0 - alpha) * eff_probs[:, ei]

    final = final / np.clip(final.sum(axis=1, keepdims=True), 1e-8, None)
    return final

best = None
rows = []

for alpha in np.arange(0.0, 1.01, 0.05):
    p = ensemble_probs_5class(mob_probs, eff_probs, alpha)
    y_pred = np.argmax(p, axis=1)

    acc = accuracy_score(y_true, y_pred)
    macro_f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)
    macro_prec = precision_score(y_true, y_pred, average="macro", zero_division=0)

    # composite objective prioritizing precision + accuracy
    score = 0.45 * macro_prec + 0.35 * acc + 0.20 * macro_f1
    rows.append((alpha, acc, macro_prec, macro_f1, score))

    if best is None or score > best[-1]:
        best = (alpha, acc, macro_prec, macro_f1, score)

print("alpha, acc, macro_precision, macro_f1, score")
for r in rows:
    print([round(x, 4) if isinstance(x, float) else x for x in r])

print("\nBEST alpha =", round(best[0], 2))
print("BEST acc =", round(best[1], 4))
print("BEST macro_precision =", round(best[2], 4))
print("BEST macro_f1 =", round(best[3], 4))
```

```python
best_alpha = best[0]
ens_prob = ensemble_probs_5class(mob_probs, eff_probs, best_alpha)
ens_pred = np.argmax(ens_prob, axis=1)

ens_acc = accuracy_score(y_true, ens_pred)
ens_macro_prec = precision_score(y_true, ens_pred, average="macro", zero_division=0)
ens_macro_f1 = f1_score(y_true, ens_pred, average="macro", zero_division=0)

print(f"[ENSEMBLE] alpha={best_alpha:.2f}")
print(f"[ENSEMBLE] Accuracy={ens_acc:.4f}")
print(f"[ENSEMBLE] Macro Precision={ens_macro_prec:.4f}")
print(f"[ENSEMBLE] Macro F1={ens_macro_f1:.4f}")

rep = classification_report(y_true, ens_pred, target_names=CLASSES_5, zero_division=0)
print("\nClassification Report:\n", rep)

cm = confusion_matrix(y_true, ens_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Greens", xticklabels=CLASSES_5, yticklabels=CLASSES_5)
plt.title(f"Ensemble Confusion Matrix (alpha={best_alpha:.2f})")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.tight_layout()
plt.savefig("/content/results/ensemble_5class_confusion_matrix.png", dpi=200)
plt.show()

with open("/content/results/ensemble_5class_report.txt", "w") as f:
    f.write(f"Best alpha: {best_alpha:.2f}\n")
    f.write(f"Accuracy: {ens_acc:.4f}\n")
    f.write(f"Macro Precision: {ens_macro_prec:.4f}\n")
    f.write(f"Macro F1: {ens_macro_f1:.4f}\n\n")
    f.write(rep)

print("[OK] Saved:")
print(" - /content/results/ensemble_5class_report.txt")
print(" - /content/results/ensemble_5class_confusion_matrix.png")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, f1_score, precision_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Required existing vars:
# model1, model2, val5, CLASSES_3, CLASSES_5, keras_image

def predict_paths(model, paths, img_size):
    probs = []
    for p in paths:
        im = keras_image.load_img(p, target_size=img_size)
        arr = keras_image.img_to_array(im) / 255.0
        arr = np.expand_dims(arr, axis=0)
        probs.append(model.predict(arr, verbose=0)[0])
    return np.array(probs)

val_paths = val5.filepaths
y_true = val5.classes

MOB_SIZE = (224, 224)   # adjust if different
EFF_SIZE = (260, 260)   # adjust if different

mob_probs = predict_paths(model1, val_paths, MOB_SIZE)  # [N,3]
eff_probs = predict_paths(model2, val_paths, EFF_SIZE)  # [N,5]

shared = [c for c in CLASSES_3 if c in CLASSES_5]
shared_eff_idx = [CLASSES_5.index(c) for c in shared]
shared_mob_idx = [CLASSES_3.index(c) for c in shared]

print("Shapes:", mob_probs.shape, eff_probs.shape, y_true.shape)
print("Shared classes:", shared)
```

```python
def ensemble_weighted_soft(mob_probs, eff_probs, alpha):
    final = eff_probs.copy()
    for mi, ei in zip(shared_mob_idx, shared_eff_idx):
        final[:, ei] = alpha * mob_probs[:, mi] + (1 - alpha) * eff_probs[:, ei]
    final = final / np.clip(final.sum(axis=1, keepdims=True), 1e-8, None)
    return final

best_soft = None
for alpha in np.arange(0.0, 1.01, 0.05):
    p = ensemble_weighted_soft(mob_probs, eff_probs, alpha)
    pred = np.argmax(p, axis=1)
    acc = accuracy_score(y_true, pred)
    mp = precision_score(y_true, pred, average="macro", zero_division=0)
    mf1 = f1_score(y_true, pred, average="macro", zero_division=0)
    score = 0.5*acc + 0.3*mp + 0.2*mf1
    if best_soft is None or score > best_soft["score"]:
        best_soft = {"alpha": float(alpha), "acc": acc, "mp": mp, "mf1": mf1, "score": score, "pred": pred, "prob": p}

print("Best Soft Voting alpha:", best_soft["alpha"])
print("Accuracy:", round(best_soft["acc"], 4), "Macro-Precision:", round(best_soft["mp"], 4), "Macro-F1:", round(best_soft["mf1"], 4))
```

```python
def ensemble_confidence_gated(mob_probs, eff_probs, threshold=0.55):
    final = eff_probs.copy()
    # EfficientNet max confidence over shared classes only
    eff_shared = eff_probs[:, shared_eff_idx]
    eff_shared_max = np.max(eff_shared, axis=1)

    for i in range(len(eff_probs)):
        for mi, ei in zip(shared_mob_idx, shared_eff_idx):
            if eff_shared_max[i] < threshold:
                # EfficientNet uncertain -> rely more on MobileNet
                final[i, ei] = 0.75 * mob_probs[i, mi] + 0.25 * eff_probs[i, ei]
            else:
                # EfficientNet confident -> rely more on EfficientNet
                final[i, ei] = 0.25 * mob_probs[i, mi] + 0.75 * eff_probs[i, ei]

    final = final / np.clip(final.sum(axis=1, keepdims=True), 1e-8, None)
    return final

best_gate = None
for t in np.arange(0.40, 0.81, 0.05):
    p = ensemble_confidence_gated(mob_probs, eff_probs, threshold=float(t))
    pred = np.argmax(p, axis=1)
    acc = accuracy_score(y_true, pred)
    mp = precision_score(y_true, pred, average="macro", zero_division=0)
    mf1 = f1_score(y_true, pred, average="macro", zero_division=0)
    score = 0.5*acc + 0.3*mp + 0.2*mf1
    if best_gate is None or score > best_gate["score"]:
        best_gate = {"t": float(t), "acc": acc, "mp": mp, "mf1": mf1, "score": score, "pred": pred, "prob": p}

print("Best Gated threshold:", best_gate["t"])
print("Accuracy:", round(best_gate["acc"], 4), "Macro-Precision:", round(best_gate["mp"], 4), "Macro-F1:", round(best_gate["mf1"], 4))
```

```python
def to_rank_probs(prob):
    # Convert per-sample class probabilities to rank-based scores [1..C], then normalize
    C = prob.shape[1]
    ranks = np.argsort(np.argsort(prob, axis=1), axis=1) + 1
    ranks = ranks.astype(np.float32) / np.sum(ranks, axis=1, keepdims=True)
    return ranks

def ensemble_rank_average(mob_probs, eff_probs, alpha=0.5):
    eff_rank = to_rank_probs(eff_probs[:, shared_eff_idx])   # [N,3]
    mob_rank = to_rank_probs(mob_probs[:, shared_mob_idx])   # [N,3]

    final = eff_probs.copy()
    shared_rank = alpha * mob_rank + (1 - alpha) * eff_rank
    shared_rank = shared_rank / np.clip(shared_rank.sum(axis=1, keepdims=True), 1e-8, None)

    for j, ei in enumerate(shared_eff_idx):
        final[:, ei] = shared_rank[:, j]

    final = final / np.clip(final.sum(axis=1, keepdims=True), 1e-8, None)
    return final

best_rank = None
for alpha in np.arange(0.0, 1.01, 0.05):
    p = ensemble_rank_average(mob_probs, eff_probs, alpha=float(alpha))
    pred = np.argmax(p, axis=1)
    acc = accuracy_score(y_true, pred)
    mp = precision_score(y_true, pred, average="macro", zero_division=0)
    mf1 = f1_score(y_true, pred, average="macro", zero_division=0)
    score = 0.5*acc + 0.3*mp + 0.2*mf1
    if best_rank is None or score > best_rank["score"]:
        best_rank = {"alpha": float(alpha), "acc": acc, "mp": mp, "mf1": mf1, "score": score, "pred": pred, "prob": p}

print("Best Rank alpha:", best_rank["alpha"])
print("Accuracy:", round(best_rank["acc"], 4), "Macro-Precision:", round(best_rank["mp"], 4), "Macro-F1:", round(best_rank["mf1"], 4))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold

# Build feature matrix:
# [eff5 probs | mob3 probs mapped to 5 dims]
mob_5 = np.zeros_like(eff_probs)
for mi, ei in zip(shared_mob_idx, shared_eff_idx):
    mob_5[:, ei] = mob_probs[:, mi]

X = np.concatenate([eff_probs, mob_5], axis=1)  # shape [N, 10]
y = y_true.copy()

# CV estimate (quick)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_pred = np.zeros_like(y)

for tr, va in skf.split(X, y):
    meta = LogisticRegression(max_iter=2000, multi_class="multinomial")
    meta.fit(X[tr], y[tr])
    oof_pred[va] = meta.predict(X[va])

stack_acc = accuracy_score(y, oof_pred)
stack_mp = precision_score(y, oof_pred, average="macro", zero_division=0)
stack_mf1 = f1_score(y, oof_pred, average="macro", zero_division=0)

print("Stacking CV Accuracy:", round(stack_acc, 4))
print("Stacking CV Macro-Precision:", round(stack_mp, 4))
print("Stacking CV Macro-F1:", round(stack_mf1, 4))

# Fit final meta-model on full val features (for deployment demo)
meta_final = LogisticRegression(max_iter=2000, multi_class="multinomial")
meta_final.fit(X, y)
```
